```python
# Step 3: Provide model.pth weights
import os, shutil

# Check if model.pth exists in Google Drive or local upload
drive_model_path = '/content/drive/MyDrive/model.pth'
if os.path.exists(drive_model_path):
    print('Copying model.pth from Google Drive...')
    shutil.copy(drive_model_path, '/content/av-forensics-flask/model.pth')
    print('✅ model.pth successfully copied!')
elif not os.path.exists('/content/av-forensics-flask/model.pth'):
    print('⚠️ model.pth not found automatically.')
    print('Please either:')
    print('1. Upload model.pth directly into /content/av-forensics-flask/ using Colab left file browser.')
    print('2. Or mount Google Drive and place model.pth in your MyDrive folder.')
else:
    print('✅ model.pth found in project directory!')
```

```python
# Step 4: Launch Flask App & Generate Live Public HTTPS URL
import subprocess, time
from pycloudflared import try_cloudflare

# Launch Flask app in background
flask_proc = subprocess.Popen(['python', 'app.py'])
time.sleep(3)

# Launch Cloudflare Tunnel to expose port 5000
tunnel = try_cloudflare(port=5000)

print('=' * 65)
print('🎉 AV-FORENSICS IS LIVE WITH GPU ACCELERATION!')
print('Open the public HTTPS link below in your browser:')
print(tunnel.tunnel_url)
print('=' * 65)
print('\nDefault Login Credentials:')
print('Username: admin')
print('Password: admin123')

# Keep cell alive while server is running
flask_proc.wait()
```
